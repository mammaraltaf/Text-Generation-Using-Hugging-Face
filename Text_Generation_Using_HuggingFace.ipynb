```python
from transformers import pipeline
from google.colab import userdata
from huggingface_hub import login

# Log in using your secure Colab secret token
hf_token = userdata.get('HF_TOKEN')
if hf_token:
    login(hf_token)

# Initialize a quick text-generation model on the GPU
# (device=0 routes the workload to your enabled T4 GPU)
generator = pipeline("text-generation", model="gpt2", device=0)

# Run an inference prompt
prompt = "The integration of AI into notebook environments allows"
output = generator(prompt, max_length=50, num_return_sequences=1)

print(output[0]['generated_text'])
```
